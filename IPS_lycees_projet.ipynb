L’enjeu de notre étude y trouve un fort intérêt : à l’aide de données recueillies par le gouvernement français centrées autour de cet indice, de différentes indications de résultats et autres statistiques, nous chercherons à comprendre, traiter puis analyser les différences, disparités et potentielles inégalités qui existent au sein des lycées français et quels en sont les déterminants. ## 1) IPS et réussite scolaire : une corrélation visible à l’échelle nationale

### A) Lien entre IPS et réussite au baccalauréat

```python
res_data = pd.read_csv('/content/fr-en-indicateurs-de-resultat-des-lycees-denseignement-general-et-technologique.csv',sep=';')
jointure = ips_lycee_2022.merge(res_data,on='UAI',how='inner')
fig, ax = plt.subplots(figsize=(45, 15))
colors = {'public':'red','privé sous contrat':'blue'}
for secteur in colors:
  subset = jointure[res_data['Secteur']==secteur]
  subset.plot.scatter(x='IPS Ensemble GT-PRO',y='Taux de mentions - Toutes series',color=colors[secteur], alpha=0.1)
  subset.plot.scatter(ax=ax,x='IPS Ensemble GT-PRO',y='Taux de mentions - Toutes series',color=colors[secteur], alpha=0.5, s=30)
```

Grâce à ces graphiques, nous pouvons voir très clairement, que ce soit pour les lycées publics ou privés, de filière générale ou technologique, qu’il existe une corrélation positive entre IPS et taux de mention au baccalauréat. Plus l’IPS est élevé, plus le nombre de lycées ayant un fort taux de mention au bac est élevé, et inversement plus l’IPS est faible plus le taux de mention est faible. Cela est facilement interprétable : les familles ayant un fort IPS sont plus à même d’aider leurs enfants et de leur permettre une bonne éducation, que ce soit en les aidant ou en leur donnant accès aux ressources nécessaires. Nous verrons par la suite plus d’explications sur ce qui influence l’IPS et donc la réussite scolaire.

### C) Capital culturel et IPS

```python
culture1 = pd.read_excel('/content/pratique culturelle en fonction de la profession du père.xlsx')
culture1.plot.bar(x='catégorie')
culture2 = pd.read_excel('/content/culture selon niveau scolaire des parents.xlsx')
culture2.plot.bar(x='niveau de diplôme des parents')
plt.show()
```

Nous savons que l’IPS est lié à la profession des parents et en lisant ces graphiques, il est clair que les enfants de cadres et professions libérales, d’artisans, commerçants et chefs d’entreprise ou de professions intérimaires (métiers demandant le plus d’années d’études) sont ceux qui ont le plus accès à la culture, que ce soit les livres ou la pratique d’activités sportives et culturelles tandis que les enfants d’agriculteurs ou d’ouvriers (demandant moins d’années d’études) semblent moins avoir accès à la culture. Or, les professions premièrement citées donnent un IPS élevé alors qu’une famille d’agriculteurs ou d’ouvriers aura un IPS relativement faible. Ainsi, l’IPS et l’accès à la culture sont positivement corrélés, plus l’IPS d’une famille est élevée, plus les enfants auront un accès exponentielle aux ressources favorisant la réussite scolaire. Cela appuie nos observations de la première sous partie : un IPS élevé implique la création d’un meilleur cadre pour le travail, ce qui favorise la réussite scolaire.

## 2) Relation entre IPS et type de lycée

### A) Lycées privés et publics

```python
ips_lycee_2022['IPS Ensemble GT-PRO'] = pd.to_numeric(ips_lycee_2022['IPS Ensemble GT-PRO'], errors='coerce')
ips_lycee_2022['ponderation'] = pd.to_numeric(ips_lycee_2022['ponderation'], errors='coerce')
ips_lycee_2022['ponderation'] = pd.to_numeric(ips_lycee_2022['ponderation'], errors='coerce')
moy = ips_lycee_2022.groupby(["Académie", "Secteur"])["ponderation"]
moy2 = pd.DataFrame(moy.sum()/ips_lycee_2022.groupby(["Académie", "Secteur"])["Effectifs Ensemble GT-PRO"].sum()).reset_index()
ips_pivot2 = moy2.pivot(index='Académie', columns='Secteur', values=0)
fig, ax = plt.subplots(figsize=(45, 5))
bar_width = 0.35
index = range(len(ips_pivot2))
public_bars2 = ax.bar(index, ips_pivot2['public'], bar_width, label='Public')
private_bars2 = ax.bar([i + bar_width for i in index], ips_pivot2['privé sous contrat'], bar_width, label='Privé')
ax.set_ylabel('IPS Moyen pondéré')
ax.set_title('IPS Moyens par Académie et par Secteur')
ax.set_xticks([i + bar_width / 2 for i in index])
ax.set_xticklabels(ips_pivot2.index)
ax.legend()
plt.show()
```

En regardant ce graphique, on voit bien que dans toutes les académies, les IPS des lycées privés sont plus élevés que ceux des lycées publics. Appuyons cela en comparant 2 cartes de la France mettant en avant les IPS par département ainsi que la tendance par département aux lycées publics ou privés.

```python
# Carte de France. Dégradé en fonction de la proportion d'élève dans le privé
ips_lycee_2022 = pd.read_csv('fr-en-ips-lycees-ap2022.csv', sep=';')
general = pd.read_csv("/content/fr-en-annuaire-education.csv", sep=";")
general["UAI"]= general["Identifiant_de_l_etablissement"]
gene = general[["UAI", "latitude", "longitude"]]
ipss = ips_lycee_2022[["UAI", "Type de lycée", "IPS voie GT", "IPS Ensemble GT-PRO", "Académie", "Secteur", "Département", "Effectifs Ensemble GT-PRO"]]
combine = pd.merge(ipss, gene, how='left', on="UAI")
combinegene = combine
metropole = combinegene[(combinegene["Académie"]!="MAYOTTE")&(combinegene["Académie"]!="GUADELOUPE")&(combinegene["Académie"]!="LA REUNION")&(combinegene["Académie"]!="GUYANE")&(combinegene["Académie"]!="MARTINIQUE")]

total_eleves_par_departement = metropole.groupby('Département')['Effectifs Ensemble GT-PRO'].sum().reset_index()
total_eleves_par_departement.columns = ['Département', 'total_eleves']

lycees_publics = metropole[metropole['Secteur'] == 'public']
# Calcul du nombre total d'élèves dans les lycées publics par département
total_eleves_publics_par_departement = lycees_publics.groupby('Département')['Effectifs Ensemble GT-PRO'].sum().reset_index()
total_eleves_publics_par_departement.columns = ['Département', 'total_eleves_publics']

ratio_df = pd.merge(total_eleves_par_departement, total_eleves_publics_par_departement, on='Département', how='left')

# Remplacer les valeurs NaN par 0 (si un département n'a pas de lycées publics, le total sera 0)
ratio_df['total_eleves_publics'].fillna(0, inplace=True)

# Calcul du ratio
ratio_df['ratio_public'] = ratio_df['total_eleves_publics'] / ratio_df['total_eleves']

# Affichage du ratio des élèves dans les lycées publics par département

ratio_df['Département'] = ratio_df['Département'].apply(lambda x: x.upper())

# Remplacement des tirets par des espaces
ratio_df['Département'] = ratio_df['Département'].replace({'-': ' '}, regex=True)
ratio_df['Département'] = ratio_df['Département'].replace({'ALPES DE HTE PROVENCE': 'ALPES DE HAUTE PROVENCE'}, regex=True)

gpd = gp.read_file("contour-des-departements.geojson")
gpd["Département"] = gpd["nom"].str.upper()
gpd['Département'] = gpd['Département'].replace({'-': ' '}, regex=True)
# Enlever les accents des noms des départements
gpd['Département'] = gpd['Département'].apply(lambda x: unidecode.unidecode(x))

gpd = pd.merge(gpd, ratio_df, how="left", on="Département")

gpd.plot(column = "ratio_public", cmap="coolwarm", legend = True)

# Carte de France. Dégradé en fonction de l'IPS moyen de chaque département

ips_lycee_2022 = pd.read_csv('fr-en-ips-lycees-ap2022.csv', sep=';')
general = pd.read_csv("/content/fr-en-annuaire-education.csv", sep=";")
general["UAI"]= general["Identifiant_de_l_etablissement"]
gene = general[["UAI", "latitude", "longitude"]]
ipss = ips_lycee_2022[["UAI", "Type de lycée", "IPS voie GT", "IPS Ensemble GT-PRO", "Académie", "Secteur", "Département"]]
combine = pd.merge(ipss, gene, how='left', on="UAI")
combinegene = combine
metropole = combinegene[(combinegene["Académie"]!="MAYOTTE")&(combinegene["Académie"]!="GUADELOUPE")&(combinegene["Académie"]!="LA REUNION")&(combinegene["Académie"]!="GUYANE")&(combinegene["Académie"]!="MARTINIQUE")]
ips_dep = metropole[["IPS Ensemble GT-PRO", "Département"]].groupby("Département").mean()
ips_dep["Département"] = ips_dep.index
ips_dep["IPS"] = ips_dep["IPS Ensemble GT-PRO"]
ips_dep.reset_index(drop=True, inplace=True)
ips_dep['Département'] = ips_dep['Département'].apply(lambda x: x.upper())

# Remplacement des tirets par des espaces
ips_dep['Département'] = ips_dep['Département'].replace({'-': ' '}, regex=True)
ips_dep['Département'] = ips_dep['Département'].replace({'ALPES DE HTE PROVENCE': 'ALPES DE HAUTE PROVENCE'}, regex=True)

gpd = gp.read_file("contour-des-departements.geojson")
gpd["Département"] = gpd["nom"].str.upper()
gpd['Département'] = gpd['Département'].replace({'-': ' '}, regex=True)
# Enlever les accents des noms des départements
gpd['Département'] = gpd['Département'].apply(lambda x: unidecode.unidecode(x))

gpd = pd.merge(gpd, ips_dep, how="left", on="Département")

gpd.plot(column = "IPS", cmap="coolwarm", legend = True)
```

Ces 2 cartes confirment bien que les lycées privés ont en général un IPS plus élevé que les lycées publics. En effet, les départements ayant la plus forte concentration de lycées privés sont celles en moyenne ayant la moyenne d’IPS la plus élevé (Bretagne par exemple ou région parisienne). Ceci est aisément interprétable : les écoles privées étant sélectives, seules les élèves ayant un certain bagage culturel et des connaissances développées peuvent accéder aux écoles privées, elles sont de surcroît plus couteuses que les écoles publiques, ce qui limite l’entrée aux familles relativement aisées. De plus, que ce soit dans les lycées publics ou privées, le taux de mention au bac est fortement lié à l’IPS : plus l’IPS est élevé plus les taux de mentions sont élevés. Les lycées ayant le plus fort IPS étant globalement privées, on peut dire que les lycées privés ont globalement plus de réussite que les publics.
